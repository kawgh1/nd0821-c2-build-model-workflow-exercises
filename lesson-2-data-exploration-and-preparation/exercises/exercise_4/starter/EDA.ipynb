```python
import pandas as pd
import seaborn as sns
import wandb

from ydata_profiling import ProfileReport
```

```python
# Create W&B and run
run = wandb.init(
    project="exercise_4",
    job_type="eda"
)
```

```python
# Fetch Artifact and load it
artifact = run.use_artifact("exercise_4/genres_mod.parquet:latest")
df = pd.read_parquet(artifact.file())
df.head()

# If this works, your W&B auth + artifact wiring is correct.
```

```python
# Generate the profiling report (expect warnings)
profile = ProfileReport(
    df,
    title="Pandas Profiling Report",
    explorative=True
)

profile
# Let it render.
# Warnings are expected — you just need to observe and note them, not fix everything.
```

```python
# Remove duplicates
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# Minimal feature engineering

# Replace missing values and concatenate:
df['title'] = df['title'].fillna('')
df['song_name'] = df['song_name'].fillna('')

df['text_feature'] = df['title'] + ' ' + df['song_name']

# Correct conceptual note:
# This feature belongs in a feature store
# If no feature store → compute at inference time, not preprocessing
# But for this exercise, you’re just demonstrating the idea
```

```python
# Quick Seaborn plot, optional, exploratory

import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(
    y=df['genre'],
    order=df['genre'].value_counts().index
)

plt.tight_layout()
plt.savefig("genre_counts.png")
```
